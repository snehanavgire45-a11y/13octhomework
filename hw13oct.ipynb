```python
l = [5, 4, 5, 4, 6, 7, 8, 4, 3, 5, 3, 4]

unique_list = []

for i in l:
    if i not in unique_list:
        unique_list.append(i)

print(unique_list)
```

```python
l = [5, 4, 5, 4, 6, 7, 8, 4, 3, 5, 3, 4]

freq_ele = {}

for i in l:
    if i in freq_ele:
        freq_ele[i] = freq_ele[i] + 1
    else:
        freq_ele[i] = 1

print(freq_ele)
```

```python
s = "I love python programming"

freq_char = {}

for i in s:
    if i in freq_char:
        freq_char[i] = freq_char[i] + 1
    else:
        freq_char[i] = 1

print(freq_char)
```
